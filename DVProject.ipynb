```python

def highest_wickets(text):
    l=text.split('--')
    return l[0]
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(highest_wickets)
```

```python
df.head(10)
```

```python
# 10 best bowling figues
df['highest_wickets']=df['highest_wickets'].astype(int)
```

```python
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
```

```python
top_bowlers.plot(kind='bar')
```
